```python
import os, sys
from pyspark.sql import SparkSession

sys.path.append(os.path.join(os.getcwd(), "/home/jovyan/work/src/pyspark_app"))

import session

spark = SparkSession.builder.getOrCreate()
url = session.jdbc_url()
props = session.jdbc_props()
```

```python
table_name = "bronze.stops"

df = spark.read.jdbc(
    url=url,
    table=table_name,
    properties=props
)

# df.show(5)
```

```python
from pyspark.sql import SparkSession
import folium
from folium.plugins import HeatMap

coords = df.select("stop_lat", "stop_lon")

coords_pd = coords.toPandas()

mapa = folium.Map(
    location=[coords_pd["stop_lat"].mean(), coords_pd["stop_lon"].mean()],
    zoom_start=13
)

heat_data = coords_pd[["stop_lat", "stop_lon"]].values.tolist()
HeatMap(heat_data).add_to(mapa)

mapa.save("/home/jovyan/work/data/output/heatmap_paradas.html")
print("Mapa salvo em heatmap_paradas.html")
```
