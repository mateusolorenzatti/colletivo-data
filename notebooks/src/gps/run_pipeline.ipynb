```python
import papermill as pm
import os
from datetime import datetime

# Lista dos notebooks na ordem do pipeline
NOTEBOOKS = [
    "./ingestion/load_gps.ipynb",
    "./processing/poins.ipynb",
    "./processing/stops.ipynb",
    "./validation/poins.ipynb",
    "./validation/stops.ipynb",
    "./results/generate_report.ipynb"
]

# Pasta de saída
OUTPUT_DIR = "output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

def run_pipeline():
    print("🚀 Iniciando pipeline de notebooks...\n")

    for nb_path in NOTEBOOKS:
        nb_name = os.path.basename(nb_path)
        nb_out = os.path.join(OUTPUT_DIR, nb_name.replace(".ipynb", "_out.ipynb"))

        print(f"Executando: {nb_name}")
        try:
            pm.execute_notebook(
                nb_path,
                nb_out,
                log_output=True,  
                progress_bar=False
            )
            print(f" Concluído: {nb_name}\n")
            
        except Exception as e:
            print(f"Erro ao executar {nb_name}: {e}")
            print("Pipeline interrompido.\n")
            break

    print("Pipeline finalizado.")

if __name__ == "__main__":
    run_pipeline()
```
